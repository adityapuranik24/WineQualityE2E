Create Test/Train Split

```python
X_train,X_test,y_train,y_test = train_test_split(data.iloc[:,:-1],data.iloc[:,-1],test_size=0.3,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scale = StandardScaler()
```

```python
X_train = scale.fit_transform(X_train)
X_test = scale.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
regression = LinearRegression()
```

```python
regression.fit(X_train,y_train)
```

```python
regression.get_params()
```

```python
reg_pred = regression.predict(X_test)
```

```python
plt.scatter(y_test, reg_pred)
```

```python
from sklearn.metrics import accuracy_score, mean_squared_error, r2_score
```

```python
r2_score(y_test, reg_pred)
```

```python
mean_squared_error(y_test, reg_pred)
```

```python
adjusted_r2 = 1- (1-mean_squared_error(y_test, reg_pred)) * (len(y_test)-1)/(len(y_test)-X_test.shape[1]-1)
adjusted_r2
```
